# Parte 1 · Entendimento do Negócio

## 1.1 Qual problema de negócio está sendo resolvido?

Uma empresa de e-commerce cresceu rápido e passou a ter **alta variação de satisfação entre clientes**: pedidos com operação parecida geram tanto promotores quanto detratores. O problema tem três camadas:

1. **Diagnóstico:** a empresa **não sabe quais fatores operacionais** (entrega, atendimento, pedido) realmente determinam se o cliente sai satisfeito ou não.
2. **Timing:** o NPS só é coletado **depois** que a jornada termina. Quando a empresa descobre que o cliente ficou insatisfeito, a experiência ruim já aconteceu e não há mais o que corrigir.
3. **Priorização:** sem saber o que pesa mais, as áreas não sabem **onde investir primeiro**.

**Em uma frase:** a empresa precisa sair de uma postura **reativa** (medir a insatisfação depois que ela aconteceu) para uma **proativa** (identificar os fatores de risco durante a jornada e agir antes do cliente virar detrator).

Traduzindo, como no CRISP-DM, a meta vaga em metas concretas:

| Meta de negócio (o "porquê") | Meta analítica (o "como medir") |
|---|---|
| Melhorar a experiência e o NPS | Identificar e ranquear os fatores operacionais associados a notas baixas |
| Agir antes da pesquisa | Encontrar **pontos de ruptura**, limites a partir dos quais o cliente quase sempre vira detrator, para servir de gatilho de alerta |
| Priorizar investimentos | Quantificar o impacto de cada fator no % de detratores, por área responsável |

## 1.2 Por que o NPS é importante para um e-commerce?

O **Net Promoter Score** vem de uma única pergunta: *"De 0 a 10, o quanto você recomendaria nossa empresa a um amigo?"*. Os clientes são divididos em **Promotores** (9–10), **Neutros** (7–8) e **Detratores** (0–6), e o NPS é `% promotores − % detratores`, indo de −100 a +100.

No e-commerce, ele é especialmente relevante porque:

- **Trocar de loja custa um clique.** Não há vínculo físico nem contrato. Um cliente insatisfeito simplesmente compra no concorrente da próxima vez.
- **A experiência é quase toda operacional.** O cliente não vê a loja nem o vendedor. Ele vê o prazo, a entrega e o atendimento, então a satisfação é um reflexo direto da operação.
- **A reputação é pública e escalável.** Avaliações em marketplaces, redes sociais e sites de reclamação amplificam cada experiência, boa ou ruim.
- **É simples e comparável.** Um único número, fácil de acompanhar ao longo do tempo e de comparar com o mercado.

## 1.3 Quais áreas se beneficiam desses insights?

| Área | Como usa os insights |
|---|---|
| **Logística** | Priorizar pedidos em risco de atraso, rever a promessa de prazo, avaliar transportadoras por atraso (não só por velocidade) |
| **Atendimento (SAC)** | Resolver no primeiro contato, criar fila prioritária para clientes com atraso ou reclamação, reduzir o tempo de resolução |
| **Customer Experience (CX)** | Criar alertas preventivos (ex.: avisar e compensar o cliente já no 1º dia de atraso, antes que a nota desabe) |
| **Produto / Tecnologia** | Rastreio em tempo real e comunicação proativa do status do pedido para evitar contatos repetidos |
| **Pricing / Comercial** | Entender se frete, desconto e parcelamento influenciam a satisfação (a análise mostrou que **quase não influenciam**, o que evita gastar em desconto para "comprar" satisfação) |
| **Marketing / CRM** | Direcionar campanhas de recompra para quem teve boa experiência e ações de recuperação para quem teve uma ruim |
| **Estratégia / Diretoria** | Priorizar investimentos pelo impacto em satisfação e retenção |

## 1.4 Reflexão: como o NPS impacta...

### ...a recompra
Cliente satisfeito volta. Cliente insatisfeito vai para o concorrente. Conquistar um cliente novo costuma custar bem mais do que manter um atual, então cada detrator gerado pela operação é receita recorrente perdida mais um custo extra de aquisição para repor esse cliente. O NPS foi criado justamente como **preditor de crescimento** por recompra e indicação (*Reichheld, "The One Number You Need to Grow", Harvard Business Review, 2003*).

> **Por que não usamos a coluna `repeat_purchase_30d` como evidência:** ao verificar a base (notebook 02, seção 9), descobrimos que ela é **exatamente `nps_score >= 8` em 100% dos pedidos**, ou seja, foi gerada a partir da própria nota. Qualquer "prova" de que detrator não recompra seria verdadeira por construção. O argumento fica apoiado na literatura, e a recomendação para a empresa é medir a recompra real, coletada de forma independente da pesquisa.

### ...o boca a boca
O NPS mede literalmente a **intenção de recomendar**. Promotores funcionam como um canal de aquisição gratuito, e detratores fazem o contrário: uma experiência ruim tende a ser mais comentada do que uma boa e, no digital, vira avaliação pública que outros consumidores leem antes de comprar. Com **84% de detratores**, a empresa hoje tem muito mais gente falando mal do que bem da marca, e isso encarece cada cliente novo.

### ...o market share em e-commerce
Market share é o resultado acumulado de **reter** (recompra) e **atrair** (boca a boca). Com barreira de troca baixa, os clientes perdidos por experiência ruim vão para a concorrência, e cada cliente perdido é share transferido para ela. Um NPS de −80 indica que a empresa está crescendo "com um balde furado": o crescimento vem de aquisição, não de lealdade, e isso é caro e frágil.

## 1.5 Indicadores de mercado que complementariam a análise

| Indicador | Por que ajuda |
|---|---|
| **Benchmark de NPS do setor** (varejo online) | Diz se −80 é um problema da empresa ou do setor. Referência geral: abaixo de 0 é crítico, acima de +50 é excelente. |
| **SLA / OTIF logístico** (% entregue no prazo e completo) | Mede o "cumprimento da promessa", que a EDA mostrou ser o fator nº 1 |
| **Prazo prometido × prazo da concorrência** | Ajuda a calibrar a promessa: prometer um prazo realista vale mais do que prometer rápido e atrasar |
| **Reputação pública** (sites de reclamação, notas em marketplaces) | Mostra o boca a boca negativo "de fora" |
| **Taxa de resolução no primeiro contato (FCR)** | Complementa o nº de contatos, que é o 3º fator mais forte |
| **Churn / LTV / CAC** | Traduz o NPS em dinheiro: quanto vale um cliente retido e quanto custa repor um perdido |
| **Preço e frete da concorrência** | Controla o efeito de preço, que internamente mostrou pouco peso |

# Parte 2 · Definição da Target

## 2.1 Qual variável representa a satisfação do cliente?

**`nps_score`**: a nota de 0 a 10 que o cliente dá na pesquisa de NPS. Para a leitura de negócio, ela é convertida em categorias:

| Categoria | Nota | Na base |
|---|---|---|
| Detrator | abaixo de 7 | 84% |
| Neutro | de 7 a menos de 9 | 11% |
| Promotor | 9 ou mais | 4% |

Seguindo a fórmula do CRISP-DM para definir a target, **"a probabilidade de que [Unidade] irá [Ação] no [Período]"**, a pergunta que orienta o projeto fica:

> **Qual a probabilidade de que um *pedido* gere um cliente *detrator* na *pesquisa de NPS aplicada ao fim da jornada*?**

- **Unidade:** o pedido (na base, 1 pedido = 1 cliente)
- **Ação:** dar nota abaixo de 7 (virar detrator)
- **Período:** o ciclo do pedido, da compra até a pesquisa aplicada após a entrega e o pós-venda (a base não traz datas, então o horizonte é "a jornada deste pedido", não um número fixo de dias)

## 2.2 Por que ela foi escolhida?

- **É a métrica que o negócio já acompanha:** o problema foi formulado em termos de NPS pela área de Experiência do Cliente.
- **Mede lealdade, não só satisfação pontual:** perguntar se o cliente *recomendaria* captura uma intenção de relacionamento futuro, não só a satisfação com um pedido.
- **Se comunica bem:** gestores entendem "% de detratores" sem precisar de estatística.

**Por que não as alternativas?**
- `csat_internal_score` também mede satisfação, mas é um score interno, menos padronizado e menos comparável com o mercado. Além disso, é medido junto com o NPS, então é **outra medida do mesmo fenômeno**, não uma causa.
- `repeat_purchase_30d` seria um **resultado de negócio** valioso, mas só se conhece 30 dias depois e é binário (não diz o quanto o cliente estava satisfeito). Nesta base, além disso, ela é **derivada da nota** (`nps_score >= 8`, notebook 02, seção 9), então não traz informação nova.

## 2.3 Em que momento da jornada essa informação é coletada?

**Depois do encerramento da jornada de compra**, ou seja, após a entrega e eventuais contatos com o atendimento.

```
Compra ──► Pagamento ──► Separação/Envio ──► Transporte ──► Entrega ──► Atendimento pós-venda ──► PESQUISA NPS
  │            │                                 │             │                │                   ▲
  order_value  installments                  atraso?       tentativas     contatos/reclamações      nps_score
  discount                                                                  tempo de resolução      csat
                                                                                                    (depois: recompra 30d)
```

Esse é o ponto central do problema: **a nota chega tarde demais para salvar aquela experiência**. Por outro lado, quase todos os fatores que a influenciam (atraso, contatos, reclamações) **acontecem antes** da pesquisa e ficam registrados durante a jornada. É isso que permite agir preventivamente.

## 2.4 Existe algum risco de usar essa variável de forma inadequada?

Sim, vários. Organizamos pelas armadilhas de definição de target vistas no curso:

**1. Vazamento de dados (leakage):** usar informações que **não existem no momento da decisão**.
- `csat_internal_score` e `repeat_purchase_30d` são medidos **junto ou depois** do NPS. Usá-los para "explicar" ou prever o NPS daria um resultado artificialmente bom e inútil na prática: quando o CSAT existe, o NPS também já existe.
- ✅ **Tratamento:** ficam fora da lista de fatores explicativos (ver `02_preparacao_dados`, seção 9).

**2. Proxy incorreto:** o NPS é um *termômetro* da experiência, não a experiência em si.
- Otimizar só o número pode levar a distorções. Exemplos: pedir "nota 10" ao cliente, aplicar a pesquisa só em quem teve boa entrega, ou dar desconto na hora da pesquisa.
- ✅ **Cuidado:** acompanhar junto com indicadores operacionais (atraso, resolução no 1º contato) e de resultado (recompra).

**3. Dado pobre ou mal definido:**
- As notas desta base têm **casas decimais** (4,4 · 6,9), ao contrário da escala inteira do NPS tradicional. A regra de corte muda o resultado: o NPS dá **−80** no corte direto e **−74** arredondando, e 6% das notas (146) mudam de categoria. Adotamos o corte direto e documentamos a escolha.
- O **dicionário de dados é ambíguo**: `delivery_time_days` aparece como "tempo total de entrega", mas em 121 pedidos o atraso é maior que ele. A coluna é o **prazo previsto** (leitura confirmada pelo professor, notebook 02, seção 5.1), então nenhum valor foi alterado. E nem toda **regra lógica entre colunas vale**: há mais reclamações do que contatos em 97% dos pedidos e a recompra é uma função exata da nota. São sinais de base sintética; por isso nenhuma coluna é usada para corrigir outra, e as análises conjuntas são lidas com essa ressalva.

**4. Viés de resposta e fairness:**
- Nem todo cliente responde à pesquisa. Quem responde costuma estar muito satisfeito ou muito insatisfeito, o que pode distorcer o retrato.
- Uma nota individual é **ruidosa**: o NPS é uma métrica de **grupo**. Usar a nota de um único cliente para decisões individuais (ex.: bonificar ou punir um atendente por uma nota) é inadequado.

**5. Correlação ≠ causalidade:**
- A análise mostra **associação** entre fatores e nota. Para afirmar que reduzir o atraso *causa* aumento do NPS, o ideal é validar com um **teste controlado** (ex.: aplicar uma ação em parte dos pedidos atrasados e comparar).

**6. Definições mutantes e drift:**
- Se a empresa mudar a pergunta, o momento ou o canal da pesquisa, a nota deixa de ser comparável com o histórico. Qualquer mudança precisa ser registrada.
